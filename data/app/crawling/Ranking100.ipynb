```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd

# 1. 타겟 페이지 URL 설정
base_url = "https://www.10000recipe.com"
list_page = base_url + "/ranking/home_new.html"  # 예시 URL

# 2. 페이지 요청
resp = requests.get(list_page)
resp.raise_for_status()  # 오류 있을 경우 예외 발생

# 3. HTML 파싱
soup = BeautifulSoup(resp.text, "html.parser")

# 4. 리스트 컨테이너 선택
ul = soup.find("ul", class_="common_sp_list_ul ea4")
if ul is None:
    raise RuntimeError("레시피 리스트 ul 태그를 찾을 수 없습니다.")

# 5. 각 레시피 카드(li)에서 <a> 태그의 href 추출
recipe_urls = []
for li in ul.find_all("li", class_="common_sp_list_li"):
    a_tag = li.find("a", href=True)
    if a_tag:
        href = a_tag["href"]
        full_url = urljoin(base_url, href)
        recipe_urls.append(full_url)

# 6. 결과 출력 (또는 저장)
for url in recipe_urls:
    print(url)

# 필요 시 → recipe_urls 를 CSV/DB로 저장하거나 후속 작업 수행
recipe_urls = pd.DataFrame(recipe_urls, columns=["Recipe URL"])
recipe_urls.to_csv("recipe_urls.csv", index=False)
```
